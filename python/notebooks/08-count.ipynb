# Lesson 8: Count letters

**The job:** Count every letter of the text, and draw the counts as bars.

**New Python:** dictionaries, dict.get(key, 0), "#" * n for a bar, counting pairs of letters.

A dictionary keeps a value for each key: here, a count for each letter. `counts.get(ch, 0)` reads the count so far, or 0 for a letter not seen yet, and the loop adds one. Multiplying a text by a number repeats it, so `"#" * 5` is a bar five long.

Decide before you run it. Then run the cell below to see.

````python
counts = {}
for ch in "ABBA":
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
````

<details><summary>Why</summary>

The loop sees A, B, B, A. Each time it adds one to that letter's count, so both end at 2. A dictionary prints its keys in the order they were first added.

</details>

## 2. Run

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

counts = {}
for ch in t:
    counts[ch] = counts.get(ch, 0) + 1

for letter in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
    n = counts.get(letter, 0)
    print(letter, n, "#" * n)
````

## 4. Modify

Count pairs of letters (bigrams) instead of single letters. Loop over the positions `range(len(t) - 1)` and count `t[i:i+2]`, in a dictionary called `pairs`.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

pairs = {}
for ch in t:
    pairs[ch] = pairs.get(ch, 0) + 1
print(pairs)
````

Run this cell to check your answer. It looks at the variables your code made.

````python
expected = {}
for i in range(len(t) - 1):
    expected[t[i:i+2]] = expected.get(t[i:i+2], 0) + 1
assert pairs == expected, "pairs should count every two-letter piece t[i:i+2]"

print("Check passed.")
````

<details><summary>A hint</summary>

Loop `for i in range(len(t) - 1):` and use `pair = t[i:i+2]` where the code used `ch`.

</details>
