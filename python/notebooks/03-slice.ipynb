# Lesson 3: Look at parts of it

**The job:** Take pieces of the text by position: the start, the end, every fifth letter.

**New Python:** indexing: t[0], t[-1], slices: t[a:b], reversing: t[::-1], step slices: t[c::w].

Positions count from 0, and negative positions count from the end. A slice `t[a:b]` takes positions a up to (not including) b. The step slice `t[c::w]` takes every w-th letter, starting at position c: it is how you pull out the columns of a grid and the slices of a Vigenère text, so it is the most useful line in this whole course.

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()   # the cleaning line from lesson 2

print(t[0], t[-1])     # the first and the last letter
print(t[:30])          # the first 30 letters
print(t[-10:])         # the last 10
print(t[::-1][:30])    # the text read backwards, its first 30 letters
print(t[0::5])         # every fifth letter, from the first
````

## 4. Modify

Make `last_word` the last three letters of `t`. (The letter is signed: these three letters are the name, in cipher.)

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
last_word = t[:3]   # these are the first three; change the slice
print(last_word)
````

Run this cell to check your answer. It looks at the variables your code made.

````python
assert last_word == t[-3:], "last_word should be the last three letters of t"

print("Check passed.")
````

<details><summary>A hint</summary>

Negative positions count from the end: `t[-3:]` starts three from the end and runs to the end.

</details>

<details><summary>A hint</summary>
