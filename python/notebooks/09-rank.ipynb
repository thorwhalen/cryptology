# Lesson 9: Rank them

**The job:** Sort the letters from most to least common, to match them with English's commonest letters.

**New Python:** sorted() with key=, lambda, tuples, and unpacking them in a loop, max() with key=, set().

`counts.items()` gives the pairs (letter, count). `sorted` puts them in order, and `key=` says what to sort by: `lambda pair: pair[1]` is a small function, written in place, that picks the count out of a pair. `lambda` is needed here, so this is where it comes in. In English the commonest letters are E, T, A, O, I and N, roughly in that order.

## 2. Run

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
counts = {}
for ch in t:
    counts[ch] = counts.get(ch, 0) + 1

ranked = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)
for ch, n in ranked[:6]:
    print(ch, n)
print("commonest:", max(counts, key=counts.get))
print("different letters:", len(set(t)))
````
