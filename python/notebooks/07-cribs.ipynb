# Lesson 7: Look for a crib or a name

**The job:** Check which expected words appear in a text, how often, and where.

**New Python:** in, str.count(), str.find(), True and False, lists of words, str.startswith().

A crib is a word you expect in the message: a name, DEAR at the start, a sign-off at the end. Once part of a cipher is undone, or to check a guessed decryption, Python can look for every word on a list at once. `in` answers yes or no, `count` says how often, and `find` says where (or -1 when the word is not there).

## 2. Run

This is the letter from lesson 1, already deciphered, so you can see what a found crib looks like. Run it and read what it prints.

````python
pt = """DEAR AGNES, THE LAMP IN THE OLD LIGHTHOUSE WAS LIT
AGAIN LAST NIGHT, THOUGH NOBODY HAS LIVED THERE
FOR TWENTY YEARS. I ROWED OUT AT DAWN AND FOUND
FRESH FOOTPRINTS ON THE STAIRS AND A LETTER ON THE
TABLE, WRITTEN IN A CODE I DO NOT KNOW. MEET ME AT
THE HARBOUR ON FRIDAY AND BRING YOUR NOTEBOOK. TOM"""
t = "".join(ch for ch in pt if ch.isalpha()).upper()

for word in ["THE", "AND", "DEAR", "LIGHTHOUSE", "TREASURE"]:
    print(word, word in t, t.count(word), t.find(word))
````

## 4. Modify

Make `found`, the list of only those words from `cribs` that appear in `t`.

````python
pt = """DEAR AGNES, THE LAMP IN THE OLD LIGHTHOUSE WAS LIT
AGAIN LAST NIGHT, THOUGH NOBODY HAS LIVED THERE
FOR TWENTY YEARS. I ROWED OUT AT DAWN AND FOUND
FRESH FOOTPRINTS ON THE STAIRS AND A LETTER ON THE
TABLE, WRITTEN IN A CODE I DO NOT KNOW. MEET ME AT
THE HARBOUR ON FRIDAY AND BRING YOUR NOTEBOOK. TOM"""
t = "".join(ch for ch in pt if ch.isalpha()).upper()
cribs = ["THE", "HARBOUR", "TREASURE", "NOTEBOOK", "PIRATE"]
found = []
for word in cribs:
    found.append(word)   # add it only when it is in t
print(found)
````

Run this cell to check your answer. It looks at the variables your code made.

````python
assert found == [w for w in cribs if w in t], "found should keep only the cribs that are in t, in the same order"

print("Check passed.")
````

<details><summary>A hint</summary>

Put `if word in t:` in front of the `append`, and indent the `append` once more.

</details>

<details><summary>A hint</summary>

Loop over every position: `for i in range(len(t)):`, and append `i` when THE starts there. As a list comprehension: `[i for i in range(len(t)) if ...]`.

</details>

<details><summary>One way to do it</summary>

```python
pt = """DEAR AGNES, THE LAMP IN THE OLD LIGHTHOUSE WAS LIT
AGAIN LAST NIGHT, THOUGH NOBODY HAS LIVED THERE
FOR TWENTY YEARS. I ROWED OUT AT DAWN AND FOUND
FRESH FOOTPRINTS ON THE STAIRS AND A LETTER ON THE

THE HARBOUR ON FRIDAY AND BRING YOUR NOTEBOOK. TOM"""
```
