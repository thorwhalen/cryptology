Everything so far used built-ins only, on purpose: you can read every line. Python also comes with a standard library, and three of its tools save real work. `Counter` counts in one line; `textwrap.wrap` cuts a text into rows; `permutations` lists every order of some columns, which turns "look" into "search": try all 720 orders of 6 columns and print only those that contain a crib. ## 2. Run

Run this cell and read what it prints.

````python
from collections import Counter
from itertools import permutations
import textwrap

ct = """HETKY ESDIN EUTTR EHHRT ISODE ONRMF HETLF EOHTT
ENAOH BURWL RLBAI ARGTN RAOHN CCEDM AOOXL EXN"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

print(Counter(t).most_common(5))     # lesson 8 and 9 in one line
print(textwrap.wrap(t, 6)[:4])       # lesson 4's rows, as a list

# Every order of the 6 columns: keep the ones that make HARBOUR appear
for order in permutations(range(6)):
    plain = "".join(t[i + k] for i in range(0, len(t), 6) for k in order)
    if "HARBOUR" in plain:
        print(order, plain[:30])
````
