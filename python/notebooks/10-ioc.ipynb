# Lesson 10: Measure: the index of coincidence

**The job:** Work out the index of coincidence, which tells one alphabet from several.

**New Python:** sum() over a loop in brackets, dict.values(), division, round().

The index of coincidence (IoC) is the chance that two letters picked at random from the text are the same. English, and anything that only relabels or moves its letters (Caesar, substitution, transposition), is about 0.066; letters spread evenly over all 26 give about 0.038. A Vigenère cipher sits in between, and its slices `t[c::w]` look like English again at the right width. The formula is: for each letter, its count n times (n - 1), all added up, divided by N times (N - 1), where N is the length.

## 2. Run

Two texts: Tom's letter (one alphabet) and a page of a logbook enciphered with a Vigenère. Run it and compare the two numbers.

````python
caesar = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
vigenere = """EHQ ZPEBTC'S XDRBADV SFDAS UC EHQ LTNFTC OR ISE
KTLR FWP LUVST IPD CXDDEP. ISE XPDT BPRE XXDTE ISE
EWTPE ISAF ELSETO, TTT HEMISED TLCT BZRZXYG, MCO A
ZDEE UC OIRUPRQCE HMCOWDXEIZV EHMI DAKH ZNXN EHUH:
EHQ ATGTI HIXA DHUCP ASPTN IWPN FWP TTGPE EIZNQH
LRQ IFRZTO. NAQZDK XY TTT GIXALGQ ZYOIH HHURS
SFDYEE PCE YTLNF, DC WTD HRAIP TTT YOFT."""

for ct in [caesar, vigenere]:
    t = "".join(ch for ch in ct if ch.isalpha()).upper()
    counts = {}
    for ch in t:
        counts[ch] = counts.get(ch, 0) + 1
    N = len(t)
    ioc = sum(n * (n - 1) for n in counts.values()) / (N * (N - 1))
    print(N, "letters, IoC", round(ioc, 3))
````

Make `slice_ioc`, the IoC of the Vigenère text's first slice at width 4, `t[0::4]`.

````python
vigenere = """EHQ ZPEBTC'S XDRBADV SFDAS UC EHQ LTNFTC OR ISE
KTLR FWP LUVST IPD CXDDEP. ISE XPDT BPRE XXDTE ISE
EWTPE ISAF ELSETO, TTT HEMISED TLCT BZRZXYG, MCO A
ZDEE UC OIRUPRQCE HMCOWDXEIZV EHMI DAKH ZNXN EHUH:
EHQ ATGTI HIXA DHUCP ASPTN IWPN FWP TTGPE EIZNQH
LRQ IFRZTO. NAQZDK XY TTT GIXALGQ ZYOIH HHURS
SFDYEE PCE YTLNF, DC WTD HRAIP TTT YOFT."""
t = "".join(ch for ch in vigenere if ch.isalpha()).upper()
s = t   # change this to the slice
counts = {}
for ch in s:
    counts[ch] = counts.get(ch, 0) + 1
N = len(s)
slice_ioc = sum(n * (n - 1) for n in counts.values()) / (N * (N - 1))
print(round(slice_ioc, 3))
````

Run this cell to check your answer. It looks at the variables your code made.

````python
_s = "".join(ch for ch in vigenere if ch.isalpha()).upper()[0::4]
_c = {}
for _ch in _s:
    _c[_ch] = _c.get(_ch, 0) + 1
_ioc = sum(n * (n - 1) for n in _c.values()) / (len(_s) * (len(_s) - 1))
assert abs(slice_ioc - _ioc) < 1e-9, "slice_ioc should be the IoC of t[0::4], not of the whole text"

print("Check passed.")
````

## 5. Make

Make `average`, the average IoC of the four slices `t[0::4]`, `t[1::4]`, `t[2::4]` and `t[3::4]` of the Vigenère text. If it is near 0.066, the key is probably 4 letters long. (A loop over the four starts, adding each IoC to a total, then divide by 4.)

````python
vigenere = """EHQ ZPEBTC'S XDRBADV SFDAS UC EHQ LTNFTC OR ISE
KTLR FWP LUVST IPD CXDDEP. ISE XPDT BPRE XXDTE ISE
EWTPE ISAF ELSETO, TTT HEMISED TLCT BZRZXYG, MCO A
ZDEE UC OIRUPRQCE HMCOWDXEIZV EHMI DAKH ZNXN EHUH:
EHQ ATGTI HIXA DHUCP ASPTN IWPN FWP TTGPE EIZNQH
LRQ IFRZTO. NAQZDK XY TTT GIXALGQ ZYOIH HHURS
SFDYEE PCE YTLNF, DC WTD HRAIP TTT YOFT."""
t = "".join(ch for ch in vigenere if ch.isalpha()).upper()
# your code here
````

Run this cell to check your answer. It looks at the variables your code made.

````python
_t = "".join(ch for ch in vigenere if ch.isalpha()).upper()
_total = 0
for _c0 in range(4):
    _s = _t[_c0::4]
    _c = {}
    for _ch in _s:
        _c[_ch] = _c.get(_ch, 0) + 1
    _total = _total + sum(n * (n - 1) for n in _c.values()) / (len(_s) * (len(_s) - 1))
assert abs(average - _total / 4) < 1e-9, "average should be the four slices' IoCs added up and divided by 4"

print("Check passed.")
````
