# Lesson 11: Letters as numbers

**The job:** Turn letters into numbers and back, so a shift is just adding and wrapping round.

**New Python:** ord() and chr(), % to wrap round the alphabet, str.index().

Every character has a number: `ord("A")` is 65, `ord("B")` is 66, and `chr` goes back. Subtract 65 and the alphabet is 0 to 25. Shifting is then adding, and `% 26` (the remainder after dividing by 26) wraps Z round to A. Caesar, affine and Vigenère all come down to this one line.

## 1. Predict

What does this print?

```python
x = ord("Y") - 65
print(chr((x + 3) % 26 + 65))
```

A. `'B'`
B. `'\\'`
C. `'AB'`
D. `'1'`

Decide before you run it. Then run the cell below to see.

````python
x = ord("Y") - 65
print(chr((x + 3) % 26 + 65))
````

<details><summary>Why</summary>

Y is number 24. Add 3 to get 27, and 27 % 26 is 1, which is B. Without the `% 26`, `chr(27 + 65)` is a backslash: the character after Z in the computer's table.

</details>

## 2. Run

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

k = 3   # shift back by k
out = ""
for ch in t[:40]:
    x = ord(ch) - ord("A")          # A=0, B=1, ... Z=25
    out = out + chr((x - k) % 26 + ord("A"))
print(t[:40])
print(out)
````

## 4. Modify

Use positions in an alphabet instead of `ord` and `chr`: with `ALPHABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"`, `ALPHABET.index(ch)` is the letter's number and `ALPHABET[x]` is the letter. Make `out`, the first 40 letters of `t` shifted back by 11.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
ALPHABET = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

k = 3
out = ""
for ch in t[:40]:
    x = ord(ch) - ord("A")
    out = out + chr((x - k) % 26 + ord("A"))
print(out)
````

Run this cell to check your answer. It looks at the variables your code made.

````python
_A = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
assert out == "".join(_A[(_A.index(c) - 11) % 26] for c in t[:40]), "out should be the first 40 letters shifted back by 11"

print("Check passed.")
````

<details><summary>A hint</summary>

`x = ALPHABET.index(ch)` and `out = out + ALPHABET[(x - k) % 26]`, with `k = 11`.

</details>
