# Strings: split, join, and views of a text

## 1. A string is a row of letters

Every letter of a string has a place, counted from 0. Square brackets pick one out; negative places count from the end.

```python
word = "CIPHER"
print(word[0], word[1], word[-1], word[-2])
print(len(word))
print("H" in word, "Z" in word)
for i, letter in enumerate(word):
    print(i, letter)
```

## 2. Slices: a piece of a string

`text[a:b]` is the piece from place `a` up to (not including) `b`. Leave out `a` to start at the beginning, `b` to go to the end. A third number is the **step**: `text[::2]` is every second letter, and `text[::-1]` reads it backwards.

```python
text = "ABCDEFGHIJKL"
print(text[2:5])      # CDE
print(text[:3])       # the first 3
print(text[-3:])      # the last 3
print(text[::2])      # every 2nd, from 0
print(text[1::2])     # every 2nd, from 1
print(text[::-1])     # backwards
print(text[2::4])     # every 4th, from 2
```

## 3. Cleaning: only the letters, in capitals

String **methods** are functions that belong to strings: `.upper()`, `.lower()`, `.replace(old, new)`, `.isalpha()` (is it a letter?), `.strip()` (cut spaces off the ends). They never change the string; they return a new one.

```python
messy = "  Meet me at noon, by the old bridge!  "
print(messy.upper())
print(messy.strip())
print(messy.replace(" ", ""))
print([ch.isalpha() for ch in "A1 b!"])
```

The first tool of every cracking notebook: keep only the letters, in capitals.

```python
def clean(text):
    """Only the letters of text, in capitals."""
    return "".join(ch for ch in text.upper() if ch.isalpha())

print(clean(messy))
```

## 4. split: one string into many

`.split()` cuts a string at the spaces and gives a list of the words. `.split(",")` cuts at commas instead, or at anything you name. `.splitlines()` cuts it into lines.

```python
sentence = "the spy met the courier at dawn"
words = sentence.split()
print(words)
print(len(words), "words; the longest is", max(words, key=len))
print("RED,GREEN,BLUE".split(","))
print("KXJ-QQP-ZZA".split("-"))
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
poem = """Roses are red
Violets are blue
Codes can be cracked
And so can you"""
lines = poem.splitlines()
print(lines)
print([line.split()[0] for line in lines])    # the first word of each line
```

## 5. join: many strings into one

`join` is split backwards: `glue.join(pieces)` puts the pieces together with `glue` between them. The glue comes first, which looks odd at first. `"".join` glues with nothing at all.

```python
pieces = ["ATTACK", "AT", "DAWN"]
print(" ".join(pieces))
print("".join(pieces))
print("-".join(pieces))
print(" ".join("SPY"))           # a string is a list of letters too
print("".join(reversed(pieces)))
```

Together they reshape a sentence: split it, change the pieces, join them back.

```python
s = "the spy met the courier at dawn"
print(" ".join(w[::-1] for w in s.split()))       # each word backwards
print(" ".join(s.split()[::-1]))                  # the words in reverse order
print("".join(w[0] for w in s.split()).upper())   # first letters
```

## 6. Chunks: blocks of the same size

Ciphertexts are often printed in blocks of five, and many ciphers work on blocks. `range(0, len(text), n)` counts 0, n, 2n, ..., the start of each block.

```python
def chunks(text, n):
    """text cut into pieces of n letters (the last one may be shorter)."""
    return [text[i:i + n] for i in range(0, len(text), n)]

def grouped(text, n=5):
    """text written in blocks of n, with spaces between, as ciphertexts are printed."""
    return " ".join(chunks(text, n))

text = clean("Meet me at noon, by the old bridge")
print(chunks(text, 4))
print(grouped(text))
print(clean(grouped(text)) == text)     # nothing lost
```

The standard library has a cousin: `textwrap.wrap(text, n)` cuts a text into lines of at most `n` characters, breaking at spaces when it can.

```python
import textwrap
print(textwrap.wrap(text, 6))
print(textwrap.wrap("the spy met the courier at dawn", 12))
```

## 7. Rows: a text as a grid

Print the chunks one under another and the text becomes a **grid**, written along the rows. Change the width and the same letters line up differently.

```python
def show_rows(text, width):
    for row in chunks(text, width):
        print(" ".join(row))

show_rows(text, 5)
print()
show_rows(text, 7)
```

## 8. Columns: every n-th letter

The columns of that grid are the letters at places 0, n, 2n, ... (the first column), 1, n+1, 2n+1, ... (the second), and so on. That is a step slice: `text[i::n]`. This view is the key to a Vigenère cipher: with the right `n`, each column was shifted by the same letter of the key.

```python
def columns(text, n):
    """text split into n columns: every n-th letter, starting at 0, 1, ..., n-1."""
    return [text[i::n] for i in range(n)]

show_rows(text, 5)
for i, col in enumerate(columns(text, 5)):
    print("column", i, col)
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Each column is a text of its own: count its letters, or shift it on its own.
for col in columns(text, 3):
    print(col, len(col), max(col, key=col.count))     # its commonest letter
```

## 9. Turning a grid with zip

`zip(a, b, c)` walks several sequences side by side and gives their first letters together, then their second, and so on. Give it the rows of a grid with `*` (which spreads a list out into separate arguments) and it hands back the columns: the grid turned on its side.

```python
print(list(zip("ABC", "DEF")))
rows = ["ABC", "DEF", "GHI"]
print(list(zip(*rows)))

def turn(rows):
    """The columns of a grid of rows (as strings). zip stops at the shortest row."""
    return ["".join(col) for col in zip(*rows)]

print(turn(rows))
print(turn(turn(rows)))       # turned twice: back where it started
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Read a grid down the columns, as a columnar transposition does.
grid = chunks(clean("WE ARE DISCOVERED FLEE AT ONCE"), 5)
print(grid)
print("".join(turn(grid)))
```

## 10. Weaving columns back into a text

After changing each column (decrypting it, say), weave them back: the first letter of each column, then the second of each, and so on. The last columns can be one letter shorter, so `zip` would drop letters; `itertools.zip_longest` keeps going and fills the gaps with `""`.

```python
from itertools import zip_longest

def weave(cols):
    """The text whose columns are cols: the inverse of columns(text, len(cols))."""
    return "".join("".join(letters) for letters in zip_longest(*cols, fillvalue=""))

cols = columns(text, 5)
print(cols)
print(weave(cols))
print(weave(cols) == text)
```

Now the view a Vigenère crack uses: split into columns, change each column with its own function, weave back. Here each column is just reversed, to show the plumbing.

```python
def per_column(text, n, change):
    """Apply change to each of the n columns of text, then weave them back."""
    return weave([change(col) for col in columns(text, n)])

print(per_column(text, 5, lambda col: col))          # change nothing: the same text
print(per_column(text, 5, lambda col: col.lower()))
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Give each column a different change: make a list of functions, one per column.
changes = [str.lower, str.upper, str.lower, str.upper, str.lower]
print(weave([f(col) for f, col in zip(changes, columns(text, 5))]))
```

## 11. Swapping letters: maketrans and translate

`str.maketrans(from, to)` makes a table: the first letter of `from` becomes the first of `to`, and so on. `text.translate(table)` swaps every letter at once. It is a whole substitution cipher in two lines. Guesses in lower case stand out from letters not yet solved.

```python
def substitute(text, old, new):
    """text with each letter of old swapped for the matching letter of new."""
    return text.translate(str.maketrans(old, new))

from string import ascii_uppercase as ALPHABET
backwards = ALPHABET[::-1]              # A<->Z, B<->Y, ... (the Atbash cipher)
secret = substitute(text, ALPHABET, backwards)
print(secret)
print(substitute(secret, ALPHABET, backwards))
print(substitute(secret, "NVG", "met"))          # a partial guess, in lower case
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# A Caesar shift as a translation: the alphabet, against the alphabet moved along by k.
k = 3
shifted = ALPHABET[k:] + ALPHABET[:k]
print(substitute(text, ALPHABET, shifted))
```
