# Lesson 13: Try a partial key

**The job:** Write your guesses into a substitution ciphertext and see what appears.

**New Python:** str.maketrans(), str.translate(), capitals and lower case to tell guesses apart.

Breaking a substitution by hand means guessing a few letters, writing them in, and looking at what the guesses make. `str.maketrans("QST", "the")` builds a key from cipher letters to plain ones; `ct.translate(key)` writes it into the whole text. Guessed letters come out in lower case and the rest stay as capitals, so you see at a glance what is still to do.

## 1. Predict

What does this print?

```python
key = str.maketrans("XQ", "th")

```

A. `'thB'`
B. `'THB'`
C. `'XQB'`
D. `'th'`

Decide before you run it. Then run the cell below to see.

````python
key = str.maketrans("XQ", "th")
print("XQB".translate(key))
````

<details><summary>Why</summary>

X becomes t and Q becomes h. B is not in the key, so it stays as it was: a capital.

</details>

## 2. Run

Agnes's reply to Tom, in a substitution cipher. The commonest three-letter word in English is THE; here QST is a good bet for it.

````python
ct = """HTLN QJD, E SLVT NTLH QST CTQQTN YJR OJRFH. QST
GJHT EP LF JCH JFT: TLGS CTQQTN PQLFHP OJN
LFJQSTN, LCWLYP QST PLDT JFT. GJRFQ QST CTQQTNP
LFH QST GJDDJFTPQ WECC IT T JN Q. WSJTVTN CEQ QST
CLDK BFTW QSEP QJJ. E WECC IT LQ QST SLNIJRN JF
ONEHLY LQ PEX. LUFTP"""

guess_cipher = "QST"
guess_plain = "the"
key = str.maketrans(guess_cipher, guess_plain)
print(ct.translate(key))
````

Run this cell to check your answer. It looks at the variables your code made.
