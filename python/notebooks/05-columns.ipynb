# Lesson 5: Read a grid by columns

**The job:** Read the columns of a grid, top to bottom, as a columnar transposition does.

**New Python:** lists, list comprehensions, columns as step slices: t[c::w], zip(*rows) (a bonus).

A columnar transposition writes the message in rows and reads it off by columns. Column c of a grid of width w is the step slice `t[c::w]`: one letter from every row. A list holds several values in order, and a list comprehension builds one in a single line: `[t[i:i+w] for i in range(0, len(t), w)]` is the list of rows.

<details><summary>Why</summary>

In rows of three the grid is ABC, DEF, GHI. `t[0::3]` takes positions 0, 3 and 6, which is the first column, ADG; then BEH and CFI.

## 2. Run

Run this cell and read what it prints.

````python
ct = """HETKY ESDIN EUTTR EHHRT ISODE ONRMF HETLF EOHTT
ENAOH BURWL RLBAI ARGTN RAOHN CCEDM AOOXL EXN"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
w = 6

rows = [t[i:i+w] for i in range(0, len(t), w)]
print(rows[:3])          # the first three rows, as a list
for c in range(w):
    print(c, t[c::w])    # each column, top to bottom
````
