## 1. A dictionary: names and values

A **dictionary** stores pairs: a *key* and its *value*. Write it with curly brackets, `key: value`, and look a value up with square brackets.

```python
ages = {"Ada": 12, "Ben": 14}
print(ages["Ada"])
ages["Cleo"] = 11        # add a new pair
ages["Ben"] = 15         # change a value
print(ages)
print(len(ages), "people")
print("Dan" in ages)     # is this key there?
```

A `for` loop over `.items()` gives you each key with its value.

```python
for name, age in ages.items():
    print(name, "is", age)
```

## 2. Counting by hand

To count, start with an empty dictionary and walk through the things one at a time. The first time you see a thing, its count is 1; after that, add 1.

```python
counts = {}
for letter in "BANANA":
    if letter in counts:
        counts[letter] = counts[letter] + 1
    else:
        counts[letter] = 1
print(counts)
```

`counts.get(letter, 0)` means "the count of `letter`, or 0 if it is not there yet". It does the `if` for you, so the loop shrinks to one line.

```python
counts = {}
for letter in "BANANA":
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
print(counts.get("Z", 0))   # never seen: 0
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Count the letters of your own name, or of a sentence. What happens to spaces?
counts = {}
for letter in "MISSISSIPPI":
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
```

## 3. A counting function

Wrap the loop in a **function**, so one line counts anything. It works on any sequence: the letters of a string, the words in a list, even numbers.

```python
def count(things):
    """A dictionary: each thing, and how many times it appears."""
    counts = {}
    for thing in things:
        counts[thing] = counts.get(thing, 0) + 1
    return counts

print(count("BANANA"))
print(count(["the", "cat", "and", "the", "hat"]))
print(count([3, 1, 3, 3]))
```

## 4. Counting pairs

Codebreakers count pairs of letters too (TH, HE, IN are the commonest in English). `zip(text, text[1:])` walks two copies of the text side by side, one letter apart, so it gives every neighbouring pair.

```python
text = "THEN"
print(list(zip(text, text[1:])))

def pairs(text):
    """Every two neighbouring letters, as two-letter strings."""
    return [a + b for a, b in zip(text, text[1:])]

print(pairs("THERE"))
print(count(pairs("THE THEATRE THEN")))
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Change pairs into triples: zip three copies of the text, each one letter further on.
def triples(text):
    return [a + b + c for a, b, c in zip(text, text[1:], text[2:])]

print(count(triples("THE THEATRE THEN")))
```

## 5. The most common

`sorted` puts things in order. Give it the dictionary's `.items()` (the pairs) and a **key**: a function that says what to sort by. Here, the count, which is the second thing in each pair, `pair[1]`. `reverse=True` puts the biggest first.

```python
counts = count("MISSISSIPPI")

def by_count(pair):
    return pair[1]

print(sorted(counts.items(), key=by_count, reverse=True))
```

A function used only once can be written in place, as a `lambda`: `lambda pair: pair[1]` is the same as `by_count`. Now the function you will use again and again: the `n` most common things, or all of them in order.

```python
def most_common(counts, n=None):
    """The (thing, count) pairs, most common first; only the first n if n is given."""
    ranked = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)
    if n is None:
        return ranked
    return ranked[:n]

print(most_common(count("MISSISSIPPI")))
print(most_common(count("MISSISSIPPI"), 2))
```

When two things have the same count, which comes first? `sorted` is *stable*: things that tie stay in the order they were in, which is the order they were first seen. Remember this: the real `Counter` does exactly the same.

## 6. Totals and shares

`sum` adds numbers up, and `.values()` gives just the counts. Dividing each count by the total gives its share: the numbers codebreakers compare with English, where E is about 12% of the letters.

```python
def total(counts):
    """How many things were counted."""
    return sum(counts.values())

def shares(counts):
    """Each thing's count as a fraction of the total (they add up to 1)."""
    t = total(counts)
    return {thing: c / t for thing, c in counts.items()}

c = count("BANANA")
print(total(c))
print(shares(c))
for letter, share in shares(c).items():
    print(letter, round(100 * share), "%")
```

That `{thing: c / t for ...}` is a **dictionary comprehension**: a loop that builds a dictionary in one line.

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Draw the counts as bars: "#" * 3 is "###".
for letter, n in most_common(count("ABRACADABRA")):
    print(letter, "#" * n)
```

## 7. Adding and taking away counts

Two counts can be added: count two texts separately, then put the counts together. Taking away keeps only what is left above zero. And `elements` turns counts back into the things themselves.

```python
def add_counts(a, b):
    """The counts of a and b together."""
    out = dict(a)                       # a copy, so a is not changed
    for thing, n in b.items():
        out[thing] = out.get(thing, 0) + n
    return out

def subtract_counts(a, b):
    """What is left of a once b is taken away, keeping only counts above zero."""
    out = {}
    for thing, n in a.items():
        left = n - b.get(thing, 0)
        if left > 0:
            out[thing] = left
    return out

def elements(counts):
    """Each thing, repeated as many times as its count."""
    return [thing for thing, n in counts.items() for _ in range(n)]

print(add_counts(count("BANANA"), count("APPLE")))
print(subtract_counts(count("BANANA"), count("NAN")))
print(elements(count("BANANA")))
```

## 8. Meet the real Counter

Everything you just built is in the standard library as `collections.Counter`. It is a dictionary that counts, and your functions have methods that match them. The cell below checks they agree: `assert` stays silent when something is true, and stops with an error when it is not.

```python
from collections import Counter

text = "MISSISSIPPI"
c = Counter(text)
print(c)

assert c == count(text)
assert c.most_common() == most_common(count(text))
assert c.most_common(2) == most_common(count(text), 2)
assert c.total() == total(count(text))
assert c + Counter("RIVER") == add_counts(count(text), count("RIVER"))
assert c - Counter("SIP") == subtract_counts(count(text), count("SIP"))
assert list(c.elements()) == elements(count(text))
print("Your functions and Counter agree.")
```

| Yours | Counter's |
|---|---|
| `count(things)` | `Counter(things)` |
| `most_common(counts, n)` | `c.most_common(n)` |
| `total(counts)` | `c.total()` |
| `add_counts(a, b)` | `a + b` |
| `subtract_counts(a, b)` | `a - b` |
| `elements(counts)` | `c.elements()` |
| `counts.get(x, 0)` | `c[x]` (a missing thing counts 0, with no error) |

`Counter` is a dictionary underneath, so everything from the first chapter works on it too. From now on, use the real one: it is faster and everyone knows it. You know what it does.

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
from collections import Counter
c = Counter("ABRACADABRA")
print(c["A"], c["Z"])        # no error for a letter that is not there
c.update("CAB")              # count some more
print(c.most_common(3))
```

## 9. Count a ciphertext

A message written in a substitution cipher (every letter swapped for another, always the same one). Count its letters: the commonest are probably E and T in disguise.

```python
from collections import Counter

ciphertext = """
HTLN QJD, E SLVT NTLH QST CTQQTN YJR OJRFH. QST
GJHT EP LF JCH JFT: TLGS CTQQTN PQLFHP OJN
LFJQSTN, LCWLYP QST PLDT JFT. GJRFQ QST CTQQTNP
LFH QST GJDDJFTPQ WECC IT T JN Q. WSJTVTN CEQ QST
CLDK BFTW QSEP QJJ. E WECC IT LQ QST SLNIJRN JF
ONEHLY LQ PEX. LUFTP
"""
letters = [ch for ch in ciphertext.upper() if ch.isalpha()]
for letter, n in Counter(letters).most_common():
    print(letter, "#" * n)
```
