# Lesson 6: Rearrange columns

**The job:** Put the letters of every row back in order with a key: the way to undo a block transposition.

**New Python:** a list of numbers as a key, indexing with each number in a list, joining letters in a new order.

A block transposition shuffles every row of the grid the same way. The key to undo it is a list of positions: `order = [2, 0, 1]` means take letter 2 of the row, then letter 0, then letter 1. A columnar transposition, written in rows and read off by columns, is undone by cutting the text into equal columns and reading across them.

## 2. Run

Run this cell and read what it prints.

````python
ct = """HETKY ESDIN EUTTR EHHRT ISODE ONRMF HETLF EOHTT
ENAOH BURWL RLBAI ARGTN RAOHN CCEDM AOOXL EXN"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
w = 6
order = [2, 0, 5, 3, 1, 4]

plain = ""
for i in range(0, len(t), w):
    row = t[i:i+w]
    fixed = "".join(row[k] for k in order)
    print(row, "->", fixed)
    plain = plain + fixed
print(plain)
````

## 4. Modify

Here two numbers of the key are in the wrong places, so the rows come out nearly right. Look at the printed rows and fix `order` so that `plain` reads as English.

````python
ct = """HETKY ESDIN EUTTR EHHRT ISODE ONRMF HETLF EOHTT
ENAOH BURWL RLBAI ARGTN RAOHN CCEDM AOOXL EXN"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
w = 6
order = [2, 0, 5, 1, 3, 4]

plain = ""
for i in range(0, len(t), w):
    row = t[i:i+w]
    fixed = "".join(row[k] for k in order)
    print(fixed)
    plain = plain + fixed
````

## 5. Make

This message was written in rows of 6 letters and read off by columns, left to right, with no key. Undo it: cut `t` into 6 equal columns of `n = len(t) // 6` letters, then read across them, row by row. Make the result `plain`.

````python
ct = """MTBUI AEFEN EHOSX TLRTD EEAEA CIOOO TOTAN HGMPW
ALHTD THTWX TDOSW HTHIX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()
n = len(t) // 6        # how many letters in each column
cols = [t[i*n:(i+1)*n] for i in range(6)]
print(cols)
# now read across: row r is cols[0][r], cols[1][r], ... cols[5][r]
````

Run this cell to check your answer. It looks at the variables your code made.

````python
assert plain == "MEETATTHEOLDBOATHOUSEATSIXANDWATCHTHELIGHTFROMTHETOPWINDOWXX", "plain should be the message read row by row: the first letter of every column, then the second letters, and so on"

print("Check passed.")
````

<details><summary>A hint</summary>

Two loops: `for r in range(n):` and, inside it, `for c in range(6):`, adding `cols[c][r]` each time. Or in one line: `plain = "".join(t[r::n] for r in range(n))`.

</details>

<details><summary>One way to do it</summary>

