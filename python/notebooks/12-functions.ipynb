# Lesson 12: Keep a tool

**The job:** Wrap a job in a function, so one line runs it on any text with any key.

**New Python:** def and return, parameters, lambda, map(), and why it needs list() or join().

A function is a tool with a name. `def shift(text, k):` says what it needs; `return` hands back the answer. Once written, `shift(t, -3)` does the whole job, and a loop can try all 26 keys in two lines. `lambda` makes a small function in place, and `map` applies a function to every letter, but lazily: it does nothing until `list(...)` or `"".join(...)` asks for the results.

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

def shift(text, k):
    out = ""
    for ch in text:
        out = out + chr((ord(ch) - 65 + k) % 26 + 65)
    return out

for k in range(26):
    print(k, shift(t[:30], -k))
````

<details><summary>One way to do it</summary>

```python
def ioc(text):
    counts = {}
    for ch in text:
        counts[ch] = counts.get(ch, 0) + 1
    N = len(text)
    return sum(n * (n - 1) for n in counts.values()) / (N * (N - 1))

print(ioc("AAB"), ioc("HELLOTHERE"))
```
