# Lesson 4: Print it in a grid

**The job:** Write the text out in rows of a fixed width, to look for patterns that line up.

**New Python:** range(start, stop, step), for loops over numbers, " ".join() to space letters out, % (the remainder), a loop inside a loop.

Printing a text in rows of a fixed width is the first look at any transposition: at the right width, the rows line up and the columns show the pattern. `range(0, len(t), w)` counts 0, w, 2w, ... so each row starts where the last one ended. A transposition keeps every letter, so the right width usually divides the length exactly: `len(t) % w` is the remainder, and 0 means it divides.

````python
t = "ABCDEFGHIJ"
for i in range(0, len(t), 4):
    print(t[i:i+4])
````

<details><summary>Why</summary>

`i` takes the values 0, 4 and 8, so the slices are t[0:4], t[4:8] and t[8:12]. The last slice runs off the end, and a slice just stops there: IJ.

## 2. Run

Run this cell and read what it prints.

````python
ct = """HETKY ESDIN EUTTR EHHRT ISODE ONRMF HETLF EOHTT
ENAOH BURWL RLBAI ARGTN RAOHN CCEDM AOOXL EXN"""
t = "".join(ch for ch in ct if ch.isalpha()).upper()

w = 6
for i in range(0, len(t), w):
    print(" ".join(t[i:i+w]))
print(len(t), "letters; remainder at width", w, "is", len(t) % w)
````
