# Lesson 2: Clean it

**The job:** Keep only the letters, in capitals, so counting and slicing see the cipher and nothing else.

**New Python:** str.upper(), str.replace(), split() and join(), for loops, if and str.isalpha().

Spaces, punctuation and line breaks get in the way of counting letters and of cutting the text into pieces. A `for` loop looks at the text one character at a time; `if ch.isalpha():` keeps a character only when it is a letter. Once that is clear, the same thing fits on one line.

## 1. Predict

What does this print?

```python

out = ""
for ch in s:
    if ch.isalpha():
        out = out + ch.upper()

```

A. `'HITOM'`
B. `'HI, TOM!'`
C. `'HI TOM'`
D. `'Hi,Tom!'`

Decide before you run it. Then run the cell below to see.

````python
s = "Hi, Tom!"
out = ""
for ch in s:
    if ch.isalpha():
        out = out + ch.upper()
print(out)
````

<details><summary>Why</summary>

The loop visits H, i, the comma, the space, T, o, m and the !. Only the five letters pass `isalpha()`, and each is added in capitals.

</details>

## 2. Run

Run this cell and read what it prints.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""

# 1. Remove every space and line break: split() cuts the text at them, join() glues the pieces
print("".join(ct.split()))

# 2. Keep only the letters, in capitals, one character at a time
t = ""
for ch in ct:
    if ch.isalpha():
        t = t + ch.upper()
print(t)
print(len(ct), "characters before,", len(t), "letters after")
````

## 4. Modify

Change the loop so it keeps letters **and** spaces, so the word breaks survive. Commas, full stops and line breaks should still go.

````python
ct = """OPLC LRYPD, ESP WLXA TY ESP ZWO WTRSESZFDP HLD WTE
LRLTY WLDE YTRSE, ESZFRS YZMZOJ SLD WTGPO ESPCP
QZC EHPYEJ JPLCD. T CZHPO ZFE LE OLHY LYO QZFYO
QCPDS QZZEACTYED ZY ESP DELTCD LYO L WPEEPC ZY ESP
ELMWP, HCTEEPY TY L NZOP T OZ YZE VYZH. XPPE XP LE
ESP SLCMZFC ZY QCTOLJ LYO MCTYR JZFC YZEPMZZV. EZX"""
t = ""
for ch in ct:
    if ch.isalpha():
        t = t + ch.upper()
print(t)
````

Run this cell to check your answer. It looks at the variables your code made.

````python
assert t == "".join(c for c in ct.upper() if c.isalpha() or c == " "), "t should hold the letters and the spaces, nothing else"

print("Check passed.")
````
