## 1. A function: a name for some steps

`def` starts a function: its name, then in brackets what it needs (its **parameters**). The indented lines are its steps. `return` hands back the answer.

```python
def double(word):
    return word + word

print(double("TOM"))
print(double("ha") + "!")
print(len(double("CIPHER")))
```

`print` only shows a value; `return` gives it back, so it can be kept, joined to something, or passed to another function. A function with no `return` gives back `None`.

```python
def shout(word):
    print(word.upper())        # shows it, gives nothing back

def louder(word):
    return word.upper()        # gives it back

a = shout("hello")
b = louder("hello")
print(a, b)
```

## 2. Arguments: defaults and names

A parameter can have a **default**, used when the call leaves it out. Calls can also name their arguments, which makes them easier to read and lets you give them in any order.

```python
def repeat(word, times=2, between=""):
    return between.join([word] * times)

print(repeat("NO"))                       # times=2, between=""
print(repeat("NO", 4))
print(repeat("NO", 3, between="-"))
print(repeat(between=" ", word="KNOCK", times=3))
```

## 3. Letters as numbers: the first tools

Codebreakers turn letters into numbers (A is 0, B is 1, ... Z is 25), do sums, and turn them back. `string.ascii_uppercase` is the alphabet, `.index` finds a letter's place, and `%` (the remainder) wraps round: 27 % 26 is 1, so one past Z is back to A.

```python
from string import ascii_uppercase as ALPHABET

def to_number(letter):
    return ALPHABET.index(letter)

def to_letter(number):
    return ALPHABET[number % 26]

print(to_number("A"), to_number("Z"))
print(to_letter(0), to_letter(25), to_letter(26), to_letter(-1))
```

Now a function built from those two: move one letter `k` places along the alphabet. Anything that is not a capital letter stays as it is.

```python
def shift(letter, k):
    if letter not in ALPHABET:
        return letter
    return to_letter(to_number(letter) + k)

print(shift("A", 3), shift("Y", 3), shift("D", -3), shift("!", 3))
```

## 4. Functions built from functions

One more step up: shift every letter of a text. `"".join(...)` glues the shifted letters back into one string (the strings workbook has a whole chapter on it).

```python
def caesar(text, k):
    """Shift every letter of text k places (letters only; spaces and punctuation stay)."""
    return "".join(shift(ch, k) for ch in text.upper())

secret = caesar("Meet me by the river", 3)
print(secret)
print(caesar(secret, -3))
```

The text in triple quotes just under `def` is a **docstring**: it says what the function does. `help(caesar)` shows it.

## 5. A function is a value

Without brackets, `caesar` is the function itself, not a call to it. You can give it another name, put it in a list, or hand it to another function, which calls it when it wants.

```python
also_caesar = caesar
print(also_caesar("ABC", 1))

def twice(f, x):
    """Call f on x, then f again on the answer."""
    return f(f(x))

print(twice(double, "AB"))
print(twice(str.upper, "quiet"))
```

```python
def each(f, things):
    """f on every thing, as a list."""
    return [f(thing) for thing in things]

print(each(len, ["A", "BEE", "SEA"]))
print(each(to_number, "HELLO"))
print(each(str.lower, ["ONE", "TWO"]))
```

## 6. lambda: a function in one line

For a small function you need only once, `lambda` writes it in place, without a name: `lambda x: x + x` is the same as `double`. After the colon comes the value it returns.

```python
print((lambda x: x + x)("AB"))
print(each(lambda w: w[0], ["ALPHA", "BRAVO", "CHARLIE"]))     # first letters
print(each(lambda ch: shift(ch, 1), "HAL"))
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Last letters, then each word backwards (w[::-1]).
words = ["ALPHA", "BRAVO", "CHARLIE"]
print(each(lambda w: w[-1], words))
```

## 7. sorted, min and max, with key=

`sorted`, `min` and `max` take a **key**: a function that says what to compare. The things themselves come back, ordered by what the key says. This is where `lambda` earns its keep.

```python
words = ["SPY", "CIPHER", "KEY", "ALPHABET", "CODE"]
print(sorted(words))                              # A to Z
print(sorted(words, key=len))                     # shortest first
print(sorted(words, key=len, reverse=True))       # longest first
print(sorted(words, key=lambda w: w[-1]))         # by last letter
print(max(words, key=len), min(words, key=len))
```

The key can be any function, even one of yours. Which shift makes `"KHOOR"` contain the most vowels?

```python
def vowels(text):
    return sum(text.count(v) for v in "AEIOU")

print(max(range(26), key=lambda k: vowels(caesar("KHOOR", -k))))
```

```python
# Playground: change anything here and run it again. Nothing you do here can break the notebook.
# Sort these by how many vowels they have, then by their second letter.
print(sorted(words, key=vowels))
```

## 8. map, filter, and comprehensions

`map(f, things)` calls `f` on each thing, and `filter(test, things)` keeps the things that pass. Both give a *promise* of the results, so wrap them in `list(...)` or `"".join(...)` to see them. A **comprehension**, `[f(x) for x in things if test(x)]`, does both in one line and is what most Python code uses.

```python
text = "Hello, World!"
print(list(map(str.upper, text)))
print("".join(filter(str.isalpha, text)))
print([ch.upper() for ch in text if ch.isalpha()])
print("".join(ch.upper() for ch in text if ch.isalpha()))
```

## 9. Functions that make functions

A function can build a new function and return it. The inner one remembers the values it was made with. Here, `make_shifter(3)` gives a function that always shifts by 3.

```python
def make_shifter(k):
    def shifter(text):
        return caesar(text, k)
    return shifter

forward3 = make_shifter(3)
back3 = make_shifter(-3)
print(forward3("ABC"), back3(forward3("ABC")))
shifters = [make_shifter(k) for k in range(5)]
print([f("A") for f in shifters])
```

The standard library does this for you: `functools.partial(f, ...)` is `f` with some arguments already filled in.

```python
from functools import partial

rot13 = partial(caesar, k=13)
print(rot13("HELLO"), rot13(rot13("HELLO")))
```

## 10. Put it together: a Caesar that cracks itself

Every piece is here. Try all 26 shifts, score each one by how English it looks, and keep the best. The score is a function, passed in as an argument, so you can swap in a better one later without touching the rest.

```python
COMMON = "ETAOINSHR"           # the commonest letters of English, most common first

def common_letters(text):
    """A rough score: how many of the text's letters are common English ones."""
    return sum(ch in COMMON for ch in text)

def crack_caesar(ciphertext, score=common_letters):
    """The shift that undoes the cipher: the one whose result scores best."""
    return max(range(26), key=lambda k: score(caesar(ciphertext, -k)))

secret = caesar("THE SECRET MEETING IS AT THE OLD TREE", 11)
k = crack_caesar(secret)
print(secret)
print(k, caesar(secret, -k))
```

Look at the chain: `to_number` and `to_letter` make `shift`, `shift` makes `caesar`, `caesar` and a score make `crack_caesar`. The cracking notebooks build their tools the same way, a little at a time.
